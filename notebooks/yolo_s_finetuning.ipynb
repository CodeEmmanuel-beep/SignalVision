```python
from ultralytics import YOLO
import torch
from pathlib import Path

device = "cuda" if torch.cuda.is_available() else "cpu"
model = YOLO("yolo26s.pt")
device
```

```python
model.train(
    data="data.yaml",
    imgsz=640,
    epochs=15,
    workers=8,
    batch=16,
    device=device,
)
```
