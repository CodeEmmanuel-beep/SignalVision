```python
yaml = """path: /kaggle/input/datasets/withus001/preprocessed-lisa-traffic-light-dataset/dataset
train: train/images
val: val/images

names:
  0: stop
  1: warning
  2: go
"""

with open("data.yaml", "w") as f:
    f.write(yaml.strip())
```

```python
model = YOLO('/kaggle/input/models/withus001/traffic-light-detection-model/pytorch/default/1/runs/detect/train/weights/best.pt')
from ultralytics import YOLO

for conf in [0.25, 0.35, 0.45, 0.55, 0.70]:
    metrics = model.val(
        data="data.yaml",
        conf=conf,
        iou=0.6,
        plots=True,
        save_json=True
    )

    print(
        f"Confidence: {conf}",
        f"Precision: {metrics.box.mp:.4f}",
        f"Recall: {metrics.box.mr:.4f}",
        f"mAP50: {metrics.box.map50:.4f}",
        f"mAP50-95: {metrics.box.map:.4f}"
    )
```
